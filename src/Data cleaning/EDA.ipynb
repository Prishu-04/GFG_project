```python
print(df.isnull().sum())
```

```python
print("Duplicate rows:", df.duplicated().sum())
 
df = df.drop_duplicates() 
```

```python
df.columns = df.columns.str.replace("\n", " ")
df.columns = df.columns.str.strip()
 
print(df.columns.tolist())
```

```python
df["Hour"] = df["Timestamp"].str.split(":").str[0].astype(int)
```

```python
df["Hour"] = df["Hour"] % 24
 
print(df[["Timestamp", "Hour"]].head())
```

"AI"  "ai" "Ai" 

```python
df["Vehicle Count"] = pd.to_numeric(df["Vehicle Count"], errors="coerce")
df["Avg Speed (km/h)"] = pd.to_numeric(df["Avg Speed (km/h)"], errors="coerce")
df["Vehicle Density (%)"] = pd.to_numeric(df["Vehicle Density (%)"], errors="coerce")
```

```python
df["Vehicle Count"] = df["Vehicle Count"].fillna(df["Vehicle Count"].median())
df["Avg Speed (km/h)"] = df["Avg Speed (km/h)"].fillna(df["Avg Speed (km/h)"].median())
df["Vehicle Density (%)"] = df["Vehicle Density (%)"].fillna(df["Vehicle Density (%)"].median())
```

```python
df["Previous_Traffic"] = df["Vehicle Count"].shift(1)

df["Previous_Traffic"] = df["Previous_Traffic"].bfill()
```

```python
print(df["Congestion Level"].unique())
print(df["Congestion Level"].value_counts())
```

```python
df_simple = df.drop(columns=["IR Presence (Lane 1-4)", "Vehicle Types Detected"])
```

```python
print("Final shape:", df_simple.shape)
print(df_simple.isnull().sum())   # should all be 0 now
print(df_simple.head())
```

```python
df_simple.to_csv(r"C:\Users\KIIT\GFG_project\data\traffic_dataset_clean.csv", index=False)
print("Saved as traffic_dataset_clean.csv")
```
