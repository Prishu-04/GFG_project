```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns 
```

```python
data = pd.read_csv(r'C:\Users\KIIT\OneDrive\urban_traffic\data\cleaned\re_cleaned.csv')
```

```python
sns.displot(data['vehicle_count'])
plt.title('Vehicle Count Distribution')
plt.tight_layout()
plt.show()
```

```python
sns.lineplot(x='hour', y='vehicle_count', hue='target_congestion', data=data)
plt.title("Average Vehicle Count by Hour")
plt.xlabel('Hour')
plt.ylabel('Vehicle Count')
plt.tight_layout()
plt.grid()
plt.show()
```

```python
sns.scatterplot(x='vehicle_count', y='avg_speed_kmh', data=data, hue='target_congestion')
plt.title('Vehicle Count vs Average Speed')
plt.xlabel('Vehicle Count')
plt.ylabel('Average Speed (km/h)')
plt.tight_layout()
plt.show()
```

```python
sns.scatterplot(x='avg_waiting_time_sec', y='vehicle_count', data=data, hue=1 )
plt.title('Vehicle Count vs Average Waiting Time')
plt.xlabel('Average Waiting Time (s)')
plt.ylabel('Vehicle Count')
plt.tight_layout()
plt.show()
```
