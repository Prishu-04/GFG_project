# Data Visualisation

```python
import pandas as pd
from matplotlib import pyplot as plt 
```

```python
data = pd.read_csv(r'C:\Users\KIIT\OneDrive\urban_traffic\data\cleaned\re_cleaned.csv')
```

```python
plt.hist(data['vehicle_count'], edgecolor='black')
plt.title('Vehicle Count Distribution')
plt.xlabel('Vehicle Count')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.plot(data['hour'], data['vehicle_count'])
plt.title('Average Vehicle Count by Hour')
plt.xlabel('Hour')
plt.ylabel('Vehicle Count')
plt.show()
```

```python
plt.bar(data['day_of_week'], data['vehicle_count'], width=0.25)
plt.title('Average Vehicle Count by Day of Week')
plt.xlabel('Day of Week')
plt.ylabel('Vehicle Count')
plt.show()
```

```python
plt.bar(data['intersection_id'], data['vehicle_count'])
plt.title('Average Vehicle Count by Intersection')
plt.xlabel('Intersection ID')
plt.ylabel('Vehicle Count')
plt.xticks(rotation=45)
plt.show()
```

```python
congestion_counts = data["target_congestion"].value_counts()

plt.bar(congestion_counts.index, congestion_counts.values)

plt.title('Congestion Level Distribution')
plt.xlabel('Congestion Level')
plt.ylabel('Number of Records')

plt.show()
```

```python
plt.scatter(data['vehicle_count'], data['avg_speed_kmh'], s=10)
plt.title('Vehicle Count vs Average Speed')
plt.xlabel('Vehicle Count')
plt.ylabel('Average Speed (km/h)')
plt.tight_layout()
plt.show()
```

```python
plt.scatter(data['avg_waiting_time_sec'], data['vehicle_count'], s=10)
plt.title('Vehicle Count vs Average Waiting Time')
plt.xlabel('Average Waiting Time (s)')
plt.ylabel('Vehicle Count')
plt.tight_layout()
plt.show()
```

```python
plt.scatter(data['queue_length_veh'], data['vehicle_count'], s=10)
plt.title('Vehicle Count vs Queue Length')
plt.xlabel('Queue Length')
plt.ylabel('Vehicle Count')
plt.tight_layout()
plt.show()
```

```python
plt.scatter(data['rain_mm_15min'], data['vehicle_count'], s=10)
plt.title('Rainfall vs Vehicle Count')
plt.xlabel('Rainfall (mm)')
plt.ylabel('Vehicle Count')
plt.tight_layout()
plt.show()
```

```python
plt.scatter(data['temperature_c'], data['vehicle_count'], s=10)
plt.title('Temperature vs Vehicle Count')
plt.xlabel('Temperature (C)')
plt.ylabel('Vehicle Count')
plt.tight_layout()
plt.show()
```

```python
plt.scatter(data['vehicle_count'], data['target_next_count'], s=10)
plt.title('Current Traffic vs Next Traffic')
plt.xlabel('Current Traffic')
plt.ylabel('Next Traffic')
plt.tight_layout()
plt.show()
```
